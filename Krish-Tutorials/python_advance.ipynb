### Class Methods and Class Variables

```python
class Car:
    base_price = 10000 ## Class Variables
    def __init__(self,windows,doors,power): #Initialization Constructor
        self.windows = windows #Instance variables
        self.doors = doors #Instance variables
        self.power = power #Instance variables
    def what_base_price(self):
        print("The base price is {}".format(self.base_price))

    @classmethod    # Decorator
    def revise_base_price(cls,inflation): #(referring to the class, user defined parameter)
        cls.base_price = cls.base_price + cls.base_price*inflation

```

```python
# To access the class variable
# Method 1: Create an instance
car1 = Car(4,5,2000)
print(car1.base_price)

# Method 2: Use the class directly
print(Car.base_price)
```

```python
# Using the class method to update the class variables
# This is the preferred way
Car.revise_base_price(0.1)
print(Car.base_price)
```

```python
# Using the instance method to update the class variable
car1 = Car
car1.revise_base_price(0.1)
print(Car.base_price)
```

### Static Methods

Static methods are faster. It is the first thing that gets initializes once the class is loaded, even before an instance is created.

```python
import datetime
now = datetime.datetime.now()
```

```python
class Car:
    base_price = 10000 ## Class Variables
    def __init__(self,windows,doors,power): #Initialization Constructor
        self.windows = windows #Instance variables
        self.doors = doors #Instance variables
        self.power = power #Instance variables
    def what_base_price(self):
        print("The base price is {}".format(self.base_price))

    @staticmethod
    def check_year():
        if now.year == 2021:
            return True
        else:
            return False
```

```python
# Calling a static method
Car.check_year()
```

### OOPS - Multiple Inheritance

```python
class A:
    def method1(self):
        print("A class method is called")
        
```

```python
class B(A): # Inheriting from class A
    def method1(self):
        print("B class method is called")

    def method2(self):
        print("B class method 2 is called")
```

```python
class C(A): # Inheriting from class A
    def method1(self):
        print("C class method is called")
```

```python
class D(B,C): # Inheriting from B & C- SInce its from multiple classes, its called multiple inheritance
    def method1(self):
        print("D class method is called")
```

```python
d = D()
d.method1()
```

```python
d.method2() # Calling methods from parent class with different names
```

```python
B.method1(d) # Calling methods with same name from parent class with same names
```

```python
A.method1(d)
```

```python
class D(B,C): # Calling the methods from parent class directly - without passing the instance of child class
    def method1(self):
        print("D class method is called")
        C.method1(self)
        B.method1(self)
        A.method1(self)
```

```python
d2 = D()
d2.method1()
```

### Public, Private and Protected

```python
# All these are public variables
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
```

```python
audi = Car(4,2,'Petrol')
audi
```

```python
dir(audi) # the public variables can be seen here at the bottom 
```

```python
# All these are protected variables. They can(should) only be modified using child/derived class
class Car:
    def __init__(self,windows,doors,enginetype):
        self._windows = windows # adding 1 underscore makes the variable protected
        self._doors = doors # adding 1 underscore makes the variable protected
        self._enginetype = enginetype # adding 1 underscore makes the variable protected
```

```python
class Truck(Car):
    def __init__(self,windows,doors,enginetype,horsepower):
        super().__init__(windows,doors,enginetype)
        self.horsepower = horsepower
```

```python
truck = Truck(4,4,'Diesel',4000)
dir(truck)
# the protected variables can be seen here at the bottom with 1 underscore as prefix
# the horsepower is a public variable and is seen without _
```

```python
# Modifying the protected variable using the child class
print(truck._doors)
truck._doors = 5
print(truck._doors)
```

```python
# All these are private variables. They can(should) not be accessed from outside the class. Also, once assigned the values can not be modified
class Car:
    def __init__(self,windows,doors,enginetype):
        self.__windows = windows # adding 2 underscore makes the variable private
        self.__doors = doors # adding 2 underscore makes the variable private
        self.__enginetype = enginetype # adding 2 underscore makes the variable private
```

```python
merc = Car(2,3,'EV')
dir(merc)
# Can see the user defined variables below as _<className>_<var_name> since they are private
```

One thing to note is that python is not strongly typecasted, so even though the variables are private and protected, they can be changed.
