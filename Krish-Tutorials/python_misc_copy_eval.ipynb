##### Difference in various copy operations - shallow, deep, assignment operator(=)

```python
# = operator points the variable to the same memory location
lst1 = [1,2,3,4]
lst2 = lst1
print(lst1, lst2) # both variables have the same values
print(hex(id(lst1)),hex(id(lst2))) # both variables are pointing to the same location
```

```python
lst2[1] = 99 # assigning a new value to the copied list
print(lst1, lst2) # reflects in both lists
```

##### Shallow Copy

```python
el_list1 = [1,2,3,4]
el_list2 = el_list1.copy() # creates a different memory location
print(el_list1, el_list2) # both variables have the same values
print(hex(id(el_list1)),hex(id(el_list2))) # both variables are pointing to the same location
```

```python
el_list2[2] = 999 # changing one of the values of the second list
print(el_list1, el_list2) # the first list remains unchanged
```

##### Shallow copy - Nested List

```python
nest_lst1 = [[1,2,3,4], [5,6,7,8]]
nest_lst2 = nest_lst1.copy()
print(nest_lst1, nest_lst2) # both variables have the same values
print(hex(id(nest_lst1)),hex(id(nest_lst2))) # both variables are pointing to different memory location
```

```python
# If we make changes to the items inside the sub-lists or nested lists i.e. elements of the collection of items, it will be reflected in both lists,
# But if we make changes to the collections as a whole i.e append or delete the sub list, it won't reflect in the copied list
nest_lst1[1][2] = 777
print(nest_lst1)
print(nest_lst2)

nest_lst1.append([2,4,6,8])
print(nest_lst1)
print(nest_lst2)

nest_lst1.remove([1,2,3,4])
print(nest_lst1)
print(nest_lst2)
```

##### Deep Copy

```python
import copy
```

```python
# In a normal list(i.e NOT a nested list) - shallow copy and deep copy work the same
d_lst1 = [1,2,3,4]
d_lst2 = copy.deepcopy(d_lst1)
print(d_lst1, d_lst2)

d_lst2[3] = 555
print(d_lst1, d_lst2)
```

```python
d_nest_lst1 = [[1,2,3,4], [5,6,7,8]]
d_nest_lst2 = copy.deepcopy(d_nest_lst1)
print(d_nest_lst1, d_nest_lst2) # both variables have the same values
print(hex(id(d_nest_lst1)),hex(id(d_nest_lst2))) # both variables are pointing to different memory location
```

```python
d_nest_lst2[0][2] = 666
print(d_nest_lst1, d_nest_lst2) # both variables have the different values
```

#### Python Eval Function - eval()
Evaluating Expressions Dynamically

```python
# Eval function evaluates python expression written as strings
print(eval("5+5"))
print((eval("5+5**2")))
```

```python
eval(input("Enter the python exp:"))
```

```python
# eval works on user defined functions
def square_num(num):
    return num**2

eval("square_num(5)")
```

##### Eval function working 
1. Parse python expression
2. Compile into byte code
3. Evaluate the python expression
4. Returns the result

```python
# How does compile() work
var = compile("5*5", "<string>", "eval")
var
```

```python
eval(var)
```

```python
# Globals
x = 10
y = 20
eval("x+y+z", {"x":x, "y":y, "z":100}) # The first {} contains the global variables, 2 ways to define those
```

```python
# Locals
eval("a+b+c",{},{"a":100,"b":20,"c":15}) # the second {} contains the local variables.
```

The eval is considered as a security risk so not to be used directly
