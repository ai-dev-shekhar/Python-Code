## Inheritance

```python
# All the class variables are public
class Car():
    def __init__(self, window, door, engine_type):
        self.windows = window
        self.doors = door
        self.engine_type = engine_type

    def drive(self):
        print("Cars are awesome")
```

```python
car = Car(4,2,'Petrol')
car.drive()
```

```python
class audi(Car): # child class - audi inheriting parent class - Car
    def __init__(self, window, door, engine_type, enableai):
        super().__init__(window, door, engine_type) # calling the parent init constructor with needed variables
        self.enableai = enableai

    def self_driving(self):
        print("Audi supports self driving")
```

```python
audiq7 = audi(5,5,'Diesel',True)
```

```python
dir(audiq7)
```

```python
print(audiq7.doors)
print(audiq7.enableai)
```

```python
audiq7.self_driving()
```

### Magic Methods in Classes
Magic methods (also known as dunder methods or special methods) are methods with double leading and trailing underscores, such as __init__, __str__, __len__, etc. These methods are not meant to be called directly by the programmer; instead, they are automatically invoked by Python in specific situations or when certain operations are performed on objects of a class.

Magic methods allow you to define how objects of your class should behave in response to various operations, such as initialization, iteration, string representation, arithmetic operations, attribute access, and more. They provide a way to overload operators and implement special behaviors for your custom objects.

```python
# All the class variables are public
# Car Blueprint
class Car():
    def __init__(self, window, door, engine_type):
        self.windows = window
        self.doors = door
        self.engine_type = engine_type

    def drive(self):
        print("Cars are awesome")
```

```python
c = Car(4,5,'Electric')
```

```python
c
```

```python
dir(c)
# All the __method__ are called magic methods
```

```python
# All the class variables are public
# Car Blueprint
class Car():
    def __init__(self, window, door, engine_type):
        self.windows = window
        self.doors = door
        self.engine_type = engine_type
    def __str__(self): # Override a magic method
        print('The object has been initialized')
    def drive(self):
        print("Cars are awesome")
```

## Iterators vs Generators

```python
# Generator - are used to create iterators
def square(n):
    for i in range(n):
        # return i**2 # a normal function return type
        yield i**2 # creating a generator
```

```python
a = square(3)
print(a)
```

```python
next(a)
```

```python
lst = [1,2,3,4]
iterable = iter(lst)
print(iterable)
```

```python
try:
    print(next(iterable))
except StopIteration:
    print("Iterator is empty")
```

## Difference
1. To create iterator we use iter() and for a generator we create a function and use yield
2. Generator uses the yield keyword. It saves and can return the local variable value
3. Generator lets us write fast and compact code in python.
4. Python iterator is much more memory efficient.

### Closures

```python
# Function inside another function - closure
# The inner function can access any variable that is passed/accessible to the parent function. There is no need to explicitly pass it to the inner function
def main_welcome(msg):

    def sub_welcome():
        print("Welcome to first message")
        print(msg)
        print("This is third message")

    return sub_welcome()
```

```python
main_welcome("This is second message")
```

```python
# Closures and Initial Decorators
def main_welcome(func):

    def sub_welcome():
        print("Welcome to first message")
        print(func([1,2,3,4,5,6]))
        print("Welcome to third message")

    return sub_welcome()
```

```python
main_welcome(len)
```

```python
# Decorators
def main_welcome(func):

    def sub_welcome():
        print("THis is first message")
        func()
        print("This is third message")
    return sub_welcome()
```

```python
@main_welcome
def new_func():
    print("This is new message")
```

```python
a = main_welcome(new_func)
```
